```python
import torch.nn as nn
import torch

class SelfAttention_v1(nn.Module):
    def __init__(self,d_in,d_out): ##d_in - input embedding dim, d_out - q,k,v dim
        super().__init__()
        self.W_query = nn.Parameter(torch.rand(d_in,d_out))
        self.W_key = nn.Parameter(torch.rand(d_in,d_out))
        self.W_value = nn.Parameter(torch.rand(d_in,d_out))
    def forward(self,x):
        keys = x @ self.W_key
        queries = x @ self.W_query
        values = x @ self.W_value

        attn_scores = queries @ keys.T 
        attn_weights = torch.softmax(
            attn_scores/keys.shape[-1]**0.05,dim=-1 ## either can use sqrt of dim or can use power .05 both are same
        )

        context_vector = attn_weights @ values

        return context_vector

```

```python
inputs = torch.tensor([
    [.43,.15,.89],
    [.55,.87,.66],
    [.57,.85,.64],
    [.22,.58,.33],
    [.77,.25,.10],
    [.05,.80,.55]
])
d_in = inputs.shape[1] 
d_out = 2
```

```python
torch.manual_seed(45)
sa_v1 = SelfAttention_v1(d_in,d_out)
print(sa_v1(inputs))
```

we can also try linear layer instead of random for stable weight initialization mechanism

```python
class SelfAttention_v2(nn.Module):
    def __init__(self,d_in,d_out,qkv_bias=False): 
        super().__init__()
        self.W_query = nn.Linear(d_in,d_out,bias=qkv_bias)
        self.W_key = nn.Linear(d_in,d_out,bias = qkv_bias)
        self.W_value = nn.Linear(d_in,d_out,bias=qkv_bias)
    def forward(self,x):
        keys = x @ self.W_key
        queries = x @ self.W_query
        values = x @ self.W_value

        attn_scores = queries @ keys.T 
        attn_weights = torch.softmax(
            attn_scores/keys.shape[-1]**0.05,dim=-1 ## either can use sqrt of dim or can use power .05 both are same
        )

        context_vector = attn_weights @ values

        return context_vector

```

```python
torch.manual_seed(789)
sa_v2 = SelfAttention_v2(d_in,d_out)
print(sa_v1(inputs))
```
